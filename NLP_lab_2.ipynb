```python
import nltk
import re
import contractions
import spacy

from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
```

```python
nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
sentences = brown.sents()

print("Total sentences:", len(sentences))
```

```python
sentences_10000 = sentences[:10000]

print("Selected sentences:", len(sentences_10000))
```

```python
texts = [" ".join(sentence) for sentence in sentences_10000]

print(texts[0])
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

cleaned_texts = [clean_text(text) for text in texts]

print(cleaned_texts[0])
```

```python
def expand_contractions(text):
    return contractions.fix(text)

expanded_texts = [expand_contractions(text) for text in cleaned_texts]

print(expanded_texts[0])
```

```python
tokenized_texts = [
    word_tokenize(text)
    for text in expanded_texts
]

print(tokenized_texts[0])
```

```python
stop_words = set(stopwords.words("english"))

print(list(stop_words)[:20])
```

```python
filtered_texts = []

for tokens in tokenized_texts:
    filtered_tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    filtered_texts.append(filtered_tokens)

print(filtered_texts[0])
```

```python
stemmer = PorterStemmer()

stemmed_texts = []

for tokens in filtered_texts:
    stemmed_tokens = [
        stemmer.stem(word)
        for word in tokens
    ]

    stemmed_texts.append(stemmed_tokens)

print(stemmed_texts[0])
```

```python
lemmatized_texts = []

for tokens in filtered_texts:
    doc = nlp(" ".join(tokens))

    lemmas = [
        token.lemma_
        for token in doc
    ]

    lemmatized_texts.append(lemmas)

print(lemmatized_texts[0])
```

```python
final_data = []

for i in range(10000):
    final_data.append({
        "original": texts[i],
        "cleaned": cleaned_texts[i],
        "expanded": expanded_texts[i],
        "tokens": tokenized_texts[i],
        "without_stopwords": filtered_texts[i],
        "stemmed": stemmed_texts[i],
        "lemmatized": lemmatized_texts[i]
    })

# First sentence ka readable output
print("ORIGINAL:")
print(final_data[0]["original"])

print("\nCLEANED:")
print(final_data[0]["cleaned"])

print("\nEXPANDED:")
print(final_data[0]["expanded"])

print("\nTOKENS:")
print(final_data[0]["tokens"])

print("\nWITHOUT STOPWORDS:")
print(final_data[0]["without_stopwords"])

print("\nSTEMMED:")
print(final_data[0]["stemmed"])

print("\nLEMMATIZED:")
print(final_data[0]["lemmatized"])
```
